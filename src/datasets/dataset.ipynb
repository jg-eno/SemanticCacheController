```python
from datasets import load_dataset

ds = load_dataset("allenai/WildChat")

ds
```

```python
train_ds = ds['train']
train_ds[0]['turn']
```

```python
unique_values = ds.unique('turn')
print(unique_values)
```

```python
single_turn = ds['train'].filter(lambda x:x["turn"] == 1)
```

```python
single_turn[0]
```

```python
single_turn_user_queries = single_turn.map(
    lambda x: {
        "conversation_id": x["conversation_id"],
        "content": next(
            msg["content"]
            for msg in x["conversation"]
            if msg["role"] == "user"
        )
    },
    remove_columns=single_turn.column_names
)
```
